```python

import nltk
from nltk.stem import PorterStemmer

# Step 1: Create a stemmer
stemmer = PorterStemmer()

# Step 2: Define words
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "running",
    "connected",
    "connection",
    "happiness"
]

# Step 3: Apply stemming
stemmed_words = []

for word in words:
    stem = stemmer.stem(word)
    stemmed_words.append(stem)

# Step 4: Display results
print("Original Words:")
print(words)

print("\nStemmed Words:")
print(stemmed_words)

# Step 5: Compare original and stemmed words
print("\nWord-by-Word Comparison:")

for original, stemmed in zip(words, stemmed_words):
    print(f"{original} -> {stemmed}")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download required datasets
nltk.download("wordnet")
nltk.download("omw-1.4")

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

# Input words
words = ["running", "studies", "playing", "cars", "better"]

# Apply lemmatization
lemmatized_words = []

for word in words:
    lemma = lemmatizer.lemmatize(word)
    lemmatized_words.append(lemma)

# Display results
print("Original Words:")
print(words)

print("\nLemmatized Words:")
print(lemmatized_words)

print("\nWord-by-Word Comparison:")
for original, lemma in zip(words, lemmatized_words):
    print(f"{original} -> {lemma}")
```

```python
words = ["running", "studies", "playing", "cars", "better"]

for word in words:
    lemma = lemmatizer.lemmatize(word, pos="v")
    print(f"{word} -> {lemma}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample sentences
sentences = [
    "I love Python",
    "I love NLP",
    "Python is amazing"
]

# Create Bag of Words vectorizer
vectorizer = CountVectorizer()

# Convert text into numerical features
bow_matrix = vectorizer.fit_transform(sentences)

# Display vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Display numerical matrix
print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
documents = [
    "I love Python",
    "I love NLP",
    "Python is amazing"
]

# Create TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Convert documents into TF-IDF values
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Display vocabulary
print("Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

# Display TF-IDF matrix
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray().round(3))
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "machine learning is useful",
    "deep learning uses neural networks",
    "machine learning uses data"
]

# Bag of Words
bow = CountVectorizer()
bow_result = bow.fit_transform(documents)

print("BAG OF WORDS")
print("Vocabulary:", bow.get_feature_names_out())
print(bow_result.toarray())

# TF-IDF
tfidf = TfidfVectorizer()
tfidf_result = tfidf.fit_transform(documents)

print("\nTF-IDF")
print("Vocabulary:", tfidf.get_feature_names_out())
print(tfidf_result.toarray().round(3))
```

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: Sample text documents
documents = [
    "Hello Buddy!!! I am learning NLP.",
    "NLP is amazing and Python is useful.",
    "I am learning Python for AI."
]

# Step 2: Clean each document
cleaned_documents = []

for text in documents:
    text = text.lower()
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    cleaned_documents.append(text)

print("Cleaned Documents:")
for document in cleaned_documents:
    print(document)

# Step 3: Tokenization
tokens = [document.split() for document in cleaned_documents]

print("\nTokens:")
for token_list in tokens:
    print(token_list)

# Step 4: Stopword removal
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

filtered_documents = []

for token_list in tokens:
    filtered_words = [
        word for word in token_list
        if word not in stop_words
    ]

    filtered_documents.append(" ".join(filtered_words))

print("\nAfter Stopword Removal:")
for document in filtered_documents:
    print(document)

# Step 5: TF-IDF vectorization
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(filtered_documents)

print("\nVocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray().round(3))

print("\nPipeline Completed Successfully!")
```
